```python
gold_df = spark.read.table("hive_metastore.nhtsa_complaints.silver")
gold_df.printSchema()
```

## 1. Modifying final schema

```python
gold_df = gold_df.select(
    F.col("CMPLID").cast("string"),
    F.col("ODINO").cast("string"),
    F.col("MFR_NAME").cast("string"),
    F.col("MAKETXT").cast("string"),
    F.col("MODELTXT").cast("string"),
    F.col("YEARTXT").cast("int"),
    F.col("CRASH").cast("string"),
    F.to_date("FAILDATE", "yyyyMMdd").alias("FAILDATE"),
    F.col("FIRE").cast("string"),
    F.col("INJURED").cast("int"),
    F.col("FATALITIES").cast("int"),
    F.col("COMPDESC").cast("string"),
    F.col("CITY").cast("string"),
    F.col("STATE").cast("string"),
    F.col("VIN").cast("string"),
    F.to_date("DATEA", "yyyyMMdd").alias("DATEA"),
    F.to_date("LDATE", "yyyyMMdd").alias("LDATE"),
    F.col("MILES").cast("int"),
    F.col("OCCURENCES").cast("int"),
    F.col("CDESCR").cast("string"),
    F.col("CMPL_TYPE").cast("string"),
    F.col("POLICE_RPT_YN").cast("string"),
    F.to_date("PURCH_DT", "yyyyMMdd").alias("PURCH_DT"),
    F.col("ORIG_OWNER_YN").cast("string"),
    F.col("ANTI_BRAKES_YN").cast("string"),
    F.col("CRUISE_CONT_YN").cast("string"),
    F.col("NUM_CYLS").cast("int"),
    F.col("DRIVE_TRAIN").cast("string"),
    F.col("FUEL_SYS").cast("string"),
    F.col("FUEL_TYPE").cast("string"),
    F.col("TRANS_TYPE").cast("string"),
    F.col("VEH_SPEED").cast("int"),
    F.col("DOT").cast("string"),
    F.col("TIRE_SIZE").cast("string"),
    F.col("LOC_OF_TIRE").cast("string"),
    F.col("TIRE_FAIL_TYPE").cast("string"),
    F.col("ORIG_EQUIP_YN").cast("string"),
    F.to_date("MANUF_DT", "yyyyMMdd").alias("MANUF_DT"),
    F.col("SEAT_TYPE").cast("string"),
    F.col("RESTRAINT_TYPE").cast("string"),
    F.col("DEALER_NAME").cast("string"),
    F.col("DEALER_TEL").cast("string"),
    F.col("DEALER_CITY").cast("string"),
    F.col("DEALER_STATE").cast("string"),
    F.col("DEALER_ZIP").cast("string"),
    F.col("PROD_TYPE").cast("string"),
    F.col("REPAIRED_YN").cast("string"),
    F.col("MEDICAL_ATTN").cast("string"),
    F.col("VEHICLES_TOWED_YN").cast("string"),
    F.to_timestamp("Data_update_timestamp").alias("Data_update_timestamp")
)

gold_df.printSchema()
```

## 2.Write to Delta with mergeSchema for schema enforcement and evolution

```python
delta_path = "/mnt/delta/gold_nhtsa"
gold_table = "hive_metastore.nhtsa_complaints.gold"

# Write Delta with schema evolution
gold_df.write.format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .save(delta_path)

# Create table if not exists
spark.sql(f"CREATE TABLE IF NOT EXISTS {gold_table} USING DELTA LOCATION '{delta_path}'")

#Verify the table
display(spark.sql(f"SELECT * FROM {gold_table} LIMIT 10"))
```
